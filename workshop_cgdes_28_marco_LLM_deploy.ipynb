## Introdução

### Requisitos de ambiente
- Python3
- Pip
- Git
- curl
- Ollama
- Docker (caso deseje "empacotar" as ferramentas em um container)

### Instalação das ferramentas em ambiente Linux

Em um terminal linux execute o seguinte comando para instalação do Python e do Git. 
```bash 
sudo apt install python3-full python3-pip git curl
```

As instalações levarão alguns minutos. Ao final, teste a instalação de cada uma das ferramentas com os seguintes comandos: 
```
python --version
pip --version
git --version
curl --version
```
Caso as instalações tenham ocorrido com sucesso, a execução de cada uma dessas instruções vai informar qual versão da ferramenta foi instalada. 


Para instalação da ferramenta Ollama, é necessário baixar o _script_ de instalação e executá-lo localmente. O comando a seguir realiza esse procedimento: 
```
curl -fsSL https://ollama.com/install.sh | sh
```

Para verificar se a instalação ocorreu corretamente, execute o seguinte comando no terminal: 
```
ollama --version
```

### Execução da ferramenta Ollama

A ferramenta Ollama pode ser executada à partir do terminal com o seguinte comando: 
```
ollama serve 
```

À partir desse momento a ferramenta Ollama estará em execução e pronta para baixar modelos e iniciar as interações com o usuário, seja pelo terminal ou através de sua API. 

Obs.: a porta padrão de comunicação da ferramenta é a porta de número 11434. Interações via API deverão utilizá-la. 

### Baixando modelos LLM para a ferramenta Ollama

Os diferentes modelos LLM disponíveis para baixar estão disponíveis na página da própria ferramenta ([https://ollama.com/models](https://ollama.com/models)). Diversos modelos (e diversas versões desses modelos) estão disponíveis aos usuários do Ollama, dentre os modelos: 
- Deepseek-r1 ([https://www.deepseek.com/](https://www.deepseek.com/))
- Gemma (Google Gemini - [https://gemini.google.com/](https://gemini.google.com/))
- QWQ (Qwen / Alibaba - [https://chat.qwen.ai/](https://chat.qwen.ai/))
- Llama (Meta / Facebook - [https://www.llama.com/](https://www.llama.com/))
- etc...

---

Para cada um dos modelos, a página oferece quais são as versões que estão disponíveis daquele modelo para serem baixadas. Por exemplo, para o modelo Deepseek-r1, estão disponíveis as versões 1.5b, 7b, 8b, 14b, 32b, 70b, 671b em que cada número representa o tamanho do modelo em bilhões de tokens que o compõe. Quanto maior for o modelo, maior é sua capacidade de resposta ao usuário mas, também, maior sua utilização de recursos computacionais. Modelos grandes podem não executar ou apresentar lentidão excessiva quando utilizados em máquinas comuns. 

![Exemplo da descrição de um modelo na página da ferramenta Ollama](exemploOllamaDeepseek.png)

---

Assim que a ferramenta estiver em execução, é possível consultar quais os modelos LLM estão disponíveis para serem utilizados através do seguinte comando: 
```
ollama ps
```

![Exemplo de Ollama vazio](ollamaVazio.png)


---

Para baixar uma versão de um modelo disponível no site do Ollama, o comando é: 
```
ollama pull <nomeModelo>:<numParametros>
```
